```python
import json
file = '/mnt/pfs_l2/jieti_team/SFT/hupeng/llm_data/kimi_style/CPT_v1_Stage2/train/train_30_part1.jsonl'
```

```python
n = 10
output_file = '/mnt/pfs_l2/jieti_team/SFT/hupeng/llm_data/kimi_style/CPT_v1_Stage2/train/split_a{}.jsonl'

data = open(file, 'r').readlines()
```

```python
ave_len = len(data) // n
for i in range(n):
    code = chr(ord('a') + i)
    new_file = output_file.format(code)
    with open(new_file, 'w', encoding='utf-8') as f:
        if i == 9:
            tmp_data = data[i*ave_len:]
        else:
            tmp_data = data[i*ave_len: (i+1)*ave_len]
        for line in tmp_data:
            f.write(json.dumps(json.loads(line.strip()), ensure_ascii=False) + '\n')
```
